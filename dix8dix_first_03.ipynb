```python
library(dplyr)
library(purrr)

group_selected_data <- function(data, grouping_vars) {
  # Step 1: Remove the columns "site_index", "sample_index", "site_name"
  data <- data %>% select(-site_index, -sample_index, -site_name)
  
  # Step 2: Define the columns to check for 'mixed' values
  check_mixed_cols <- c(
    "sum", "time_range", "geological_context", "age",
    "gender", "accumulation_cause", "hominin_species"
  )
  
  # Step 3: Group data by the specified grouping variables
  data_grouped <- data %>% group_by(across(all_of(grouping_vars)))
  
  # Step 4: Summarize the data according to the specified conventions
  data_summarised <- data_grouped %>% summarise(
    across(all_of(check_mixed_cols), ~ if(n_distinct(.x) == 1) first(.x) else "mixed"),
    bone.level = list(first(bone.level)),
    number_of_individuals = sum(number_of_individuals),
    count_combined = list(reduce(count, ~ map2(.x, .y, `+`))),
  .groups = 'drop'  
  )
  
  # Ungroup the data if needed
  #data_summarised <- data_summarised %>% ungroup()
  
  return(data_summarised)
}
```

```python
library(dplyr)
library(purrr)
library(tidyr)

process_archaeological_data <- function(data, group_cols) {
  # Define columns to check for mixed values
  check_mixed_cols <- c("sum", "time_range", "geological_context", "age",
                       "gender", "accumulation_cause", "hominin_species")
  
  result <- data %>%
    #group_by(across(all_of(group_cols))) %>%
    summarise(
      # Handle mixed columns
      across(all_of(check_mixed_cols), 
             ~ if (n_distinct(.x) == 1) first(.x) else "mixed"),
      
      # Sum the number of individuals
      number_of_individuals = sum(number_of_individuals, na.rm = TRUE),
      
      # Keep first bone level
      bone.level = list(first(bone.level)),
      
      # Sum the counts
      count = list(reduce(count, ~ map2(.x, .y, `+`))),
      
      # Count number of samples in group
      sample_count = n(),
      
      .groups = "drop"
    )
  
  return(result)
}

# Example usage:
# 1. Group by site_name only
# result1 <- process_archaeological_data(data, c("site_name"))

# 2. Group by multiple columns
# result2 <- process_archaeological_data(data, c("site_name", "geological_context"))

# Helper function to inspect results
print_group_summary <- function(data, group_result) {
  cat("Original rows:", nrow(data), "\n")
  cat("Grouped rows:", nrow(group_result), "\n")
  cat("Group columns used:", paste(group_cols, collapse = ", "), "\n")
  cat("\nSample result row:\n")
  print(group_result[1,])
}
```

```python
# Load necessary libraries
library(dplyr)
library(purrr)

# Define a function to dynamically group and summarize the data
group_and_summarize <- function(data, group_cols) {
  
  # Define the columns to check for mixed values, excluding group_cols
  check_mixed_cols <- c("sum", "time_range", "geological_context", 
                        "accumulation_cause", "gender", "age", "hominin_species") %>%
                      setdiff(group_cols)
  
  # Perform the grouping and merging
  merged_df <- data %>%
    group_by(across(all_of(group_cols))) %>%
    summarise(
      # For specified columns, keep the value if unique, else mark as "mixed"
      across(all_of(check_mixed_cols), ~ if (n_distinct(.x) == 1) first(.x) else "mixed"),
      # Sum the number of individuals
      number_of_individuals = sum(number_of_individuals, na.rm = TRUE),
      # Keep the first bone.level as a list
      bone.level = list(first(bone.level)),
      # Keep the first element as a list
      element = list(first(element)),
      # Sum the 'count' lists element-wise across the group
      count = list(reduce(count, ~ map2(.x, .y, `+`))),
      .groups = 'drop'  # Ungroup after summarising (optional)
    )
  
  return(merged_df)
}

# Example of how to use the function
# Specify the grouping columns (you can modify these)
group_cols <- c("age", "hominin_species")

# Assuming 'recalc_same_data' is your input tibble/dataframe
result <- group_and_summarize(recalc_same_data, group_cols)

# View the result
print(result)
```
