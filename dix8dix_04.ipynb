```python
# Load necessary libraries
library(dplyr)
library(purrr)

# Define a function to dynamically group and summarize the data
group_and_summarize <- function(data, group_cols) {
  
  # Define the columns to check for mixed values, excluding group_cols
  check_mixed_cols <- c("sum", "time_range", "geological_context", 
                        "accumulation_cause", "gender", "age", "hominin_species") %>%
                      setdiff(group_cols)
  
  # Perform the grouping and merging
  merged_df <- data %>%
    group_by(across(all_of(group_cols))) %>%
    summarise(
      # For specified columns, keep the value if unique, else mark as "mixed"
      across(all_of(check_mixed_cols), ~ if (n_distinct(.x) == 1) first(.x) else "mixed"),
      # Sum the number of individuals
      number_of_individuals = sum(number_of_individuals, na.rm = TRUE),
      # Keep the first bone.level as a list
      bone.level = list(first(bone.level)),
      # Keep the first element as a list
      element = list(first(element)),
      # Sum the 'count' lists element-wise across the group
      count = list(reduce(count, ~ map2(.x, .y, `+`))),
      .groups = 'drop'  # Ungroup after summarising (optional)
    )
  
  return(merged_df)
}

# Example of how to use the function
# Specify the grouping columns (you can modify these)
group_cols <- c("age", "hominin_species")

# Assuming 'recalc_same_data' is your input tibble/dataframe
result <- group_and_summarize(recalc_same_data, group_cols)

# View the result
print(result)
```

```python
process_archaeological_data <- function(
  data,
  time_range_filter = NULL,
  geological_context_filter = NULL,
  number_of_individuals_filter = NULL,
  age_filter = NULL,
  gender_filter = NULL,
  accumulation_cause_filter = NULL,
  bone_parts_filter = NULL,
  site_name_filter = NULL,
  hominin_species_filter = NULL
) {
  # First, apply the select_data function to get initially filtered data
  filtered_data <- select_data(
    data = data,
    time_range_filter = time_range_filter,
    geological_context_filter = geological_context_filter,
    number_of_individuals_filter = number_of_individuals_filter,
    age_filter = age_filter,
    gender_filter = gender_filter,
    accumulation_cause_filter = accumulation_cause_filter,
    bone_parts_filter = bone_parts_filter,
    site_name_filter = site_name_filter,
    hominin_species_filter = hominin_species_filter
  )
  
  # Define columns to check for mixed values
  check_mixed_cols <- c(
    "sum", "time_range", "geological_context",
    "accumulation_cause", "gender", "age", "hominin_species"
  )
  
  # Split the data into sum=1 and sum=0 groups
  sum_1_data <- filtered_data %>% filter(sum == 1)
  sum_0_data <- filtered_data %>% filter(sum == 0)
  
  # Process sum=0 data: group by site_index and summarize
  if (nrow(sum_0_data) > 0) {
    summarized_data <- sum_0_data %>%
      group_by(site_index) %>%
      summarise(
        # For sample_index, use the first one with site_ prefix
        sample_index = paste0(first(str_extract(site_index, "\\d+")), str_extract(first(sample_index), "\\d+")),
        # For site_name, use the first one
        site_name = first(site_name),
        # Set sum to 1 as we're summarizing
        sum = 1,
        # For specified columns, keep the value if unique, else mark as "mixed"
        across(all_of(check_mixed_cols), ~ if (n_distinct(.x) == 1) first(.x) else "mixed"),
        # Sum the number of individuals
        number_of_individuals = sum(number_of_individuals, na.rm = TRUE),
        # Keep the first bone.level as a list
        bone.level = list(first(bone.level)),
        # Keep the first element as a list
        element = list(first(element)),
        # Sum the 'count' lists element-wise across the group
        count = list(reduce(count, ~ map2(.x, .y, `+`))),
        .groups = 'drop'  # Ungroup after summarising
      )
    
    # Combine the summarized data with the original sum=1 data
    final_data <- bind_rows(sum_1_data, summarized_data)
  } else {
    final_data <- sum_1_data
  }
  
  # Return the processed data
  return(final_data)
}
```

```python
library(dplyr)
library(purrr)

process_data <- function(data) {
  # Define the columns to check for mixed values
  check_mixed_cols <- c("sum", "time_range", "geological_context",
                        "accumulation_cause", "gender", "age", "hominin_species")
  
  # Split data into sum=0 and sum=1 subsets
  data_sum0 <- data %>% filter(sum == 0)
  data_sum1 <- data %>% filter(sum == 1)
  
  # If there is sum=0 data, process it
  if(nrow(data_sum0) > 0) {
    # Summarize the sum=0 data
    data_sum0_summarized <- data_sum0 %>%
      group_by(site_index) %>%
      summarise(
        # For specified columns, keep the value if unique, else mark as "mixed"
        across(all_of(check_mixed_cols), ~ if (n_distinct(.x) == 1) first(.x) else "mixed"),
        # Sum the number of individuals
        number_of_individuals = sum(number_of_individuals, na.rm = TRUE),
        # Keep the first bone.level as a list
        bone.level = list(first(bone.level)),
        # Keep the first element as a list
        element = list(first(element)),
        # Sum the 'count' lists element-wise across the group
        count = list(reduce(count, ~ map2(.x, .y, `+`))),
        .groups = 'drop'  # Ungroup after summarising
      ) %>%
      mutate(sum = 1)  # Update sum to 1
    
    # Combine the summarized sum=1 data with the original sum=1 data
    data_combined <- bind_rows(data_sum1, data_sum0_summarized)
  } else {
    # If no sum=0 data, just return data_sum1
    data_combined <- data_sum1
  }
  
  return(data_combined)
}
```
